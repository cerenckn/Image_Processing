```python
#1.fotoğrafı okuyup gri tonlamaya çevirmek 
import cv2
import numpy as np
import matplotlib.pyplot as plt
img = cv2.imread('kendi_foto.jpeg')
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.imshow(gray_img, cmap='gray')
plt.title("Gri Tonlamalı Fotoğraf")
plt.axis('off')
plt.show()
```

```python
# 2. Matris boyutlarını yazdır
yukseklik, genislik = gray_img.shape
print(f"Görüntü Boyutları: {yukseklik} x {genislik} piksel")
```

```python
# 3. ROI Slicing 
# face_roi = gray_img[y1:y2, x1:x2]
face_roi = gray_img[60:200, 200:315] 
plt.imshow(face_roi, cmap='gray')
plt.title("Kesilen Yüz Bölgesi (ROI)")
plt.axis('off')
plt.show()
cv2.waitKey(0)
```

```python
# 4. 2-bit (4 renk) Nicemleme: normalde resimler 8 bit üzerinden, 2^8=256 skalada değişiyordu.
#biz 4 bit üzerinden yapacağımız için bu da false contourşnge neden olacaktır..
bit_seviyesi = 2
renk_sayisi = 2 ** bit_seviyesi # 4
aralik = 256 // renk_sayisi # 64

# Numpy matris matematiği ile nicemleme
# burada 256/4=64 olduğundan dolayı renkler 0-63,64-127.. şeklinde aralıklarda değişir.
quantized_face = (face_roi // aralik) * aralik

plt.figure(figsize=(10,5))
plt.subplot(1, 2, 1)
plt.imshow(face_roi, cmap='gray')
plt.title("Orijinal ROI")

plt.subplot(1, 2, 2)
plt.imshow(quantized_face, cmap='gray')
plt.title("2-bit Nicemlenmiş (False Contouring)")
plt.show()
```
