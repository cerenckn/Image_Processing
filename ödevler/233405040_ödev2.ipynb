```python
#CEREN ÇEKEN- 233405040 
#VİGNETTE-AFFİNE TRANSFORM 
import cv2
import numpy as np

#fotoğrafı okuma 
img = cv2.imread('kendi_foto.jpeg')

if img is None:
    print("Resim yüklenemedi, lütfen dosya yolunu kontrol edin!")
else:

    yukseklik, genislik = img.shape[:2]
    merkez_x, merkez_y = genislik // 2, yukseklik // 2

#fotoğrafı siyah beyaz boyutlama 
    mask = np.zeros((yukseklik, genislik), dtype="uint8")
    yaricap = int(min(merkez_x, merkez_y) * 0.8)
    cv2.circle(mask, (merkez_x, merkez_y), yaricap, 255, -1)

#maskeleme
    odaklanmis_resim = cv2.bitwise_and(img, img, mask=mask)
    
#döndürme işlemi -15.derece 
    M = cv2.getRotationMatrix2D((merkez_x, merkez_y), -15, 1.0)
    sonuc = cv2.warpAffine(odaklanmis_resim, M, (genislik, yukseklik), flags=cv2.INTER_LINEAR)
    
#sonuçlar ve başka ekranda sonuç olarak gösterme 
    cv2.imshow("Sonuc", sonuc)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
